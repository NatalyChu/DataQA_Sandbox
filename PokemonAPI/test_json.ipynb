```python
import pandas as pd
import json
from sqlalchemy import create_engine
```

```python
# Параметры подключения к локальной БД
db_name = "postgres"
user = "postgres"
password = "123456"
host = "localhost"  
port = "5432"  # Стандартный порт PostgreSQL
```

```python
# Создаём подключение через SQLAlchemy
engine = create_engine(f"postgresql://{user}:{password}@{host}:{port}/{db_name}")
```

```python
df = pd.read_sql("SELECT * FROM silver.types", con=engine)

# Выводим первые 5 строк
print(df.head())
```

```python
# Путь к JSON файлу
json_file = 'type.json'

# Чтение JSON файла
with open(json_file, 'r') as f:
    data = json.load(f)

# Разворачиваем 'results'
results = data['results']
df = pd.DataFrame(results)

# Добавляем автоинкремент id (если нужно)
df.reset_index(inplace=True)
df.rename(columns={'index': 'id'}, inplace=True)
df['id'] = df['id'] + 1  # чтобы id начинался с 1

print(df)
# df.drop('id', axis=1, inplace=True)

# Запись в таблицу
df.to_sql('types_json', engine, schema='silver', if_exists='append', index=False)

print("Данные успешно загружены в PostgreSQL!")
```

```python
with open('type.json') as f:
    data = json.load(f)

# Просто оборачиваем весь 'results' как json строку
df = pd.DataFrame([{'raw_data': json.dumps(data['results'])}])

# Загружаем в БД
df.to_sql('types_json_raw', engine, schema='silver', if_exists='append', index=False)
```
